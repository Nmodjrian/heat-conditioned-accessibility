```r
library(spanishoddata)
library(dplyr)
library(sf)

project_root <- paste0(
  "/scratch/project_2010417/paper3/",
  "heat-conditioned-accessibility"
)

mobile_raw <- file.path(
  project_root,
  "data/mobile/raw"
)

mobile_interim <- file.path(
  project_root,
  "data/mobile/interim"
)

dir.create(
  mobile_raw,
  recursive = TRUE,
  showWarnings = FALSE
)

dir.create(
  mobile_interim,
  recursive = TRUE,
  showWarnings = FALSE
)

spod_set_data_dir(mobile_raw)

spod_get_data_dir()
```

```r
districts <- spod_get_zones(
  zones = "districts",
  ver = 2
)

names(districts)
```

```r
sevilla_candidates <- districts |>
  filter(
    grepl(
      "Sevilla",
      name,
      ignore.case = TRUE
    )
  )

sevilla_candidates
```

```r
st_write(
  districts,
  file.path(
    mobile_interim,
    "spain_districts_v2.gpkg"
  ),
  delete_dsn = TRUE
)
```

```r
names(districts)
```

```r
sevilla_candidates
```

```r
sevilla_zones <- districts |>
  filter(municipalities == "41091")

sevilla_zones |>
  select(
    id,
    name,
    population,
    municipalities,
    luas_mitma
  )
```

```r
st_write(
  sevilla_zones,
  file.path(
    mobile_interim,
    "sevilla_districts_v2.gpkg"
  ),
  delete_dsn = TRUE
)
```

```r
sevilla_ids <- sevilla_zones$id

sevilla_ids
```

```r
test_date <- "2024-07-15"
```

```r
od_test <- spod_get(
  type = "od",
  zones = "districts",
  dates = test_date
)
```

```r
class(od_test)

names(od_test)
```

```r
sevilla_od_test <- od_test |>
  filter(
    id_origin %in% sevilla_ids |
    id_destination %in% sevilla_ids
  )
```

```r
sevilla_od_test |>
  head()
```

```r
sevilla_od_test <- sevilla_od_test |>
  collect()

dim(sevilla_od_test)

names(sevilla_od_test)

head(sevilla_od_test)
```

```r
arrow::write_parquet(
  sevilla_od_test,
  file.path(
    mobile_interim,
    "sevilla_od_2024-07-15.parquet"
  )
)
```
