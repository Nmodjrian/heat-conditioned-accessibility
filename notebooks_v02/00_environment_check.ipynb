# 00 — Environment check

This notebook is a lightweight smoke test for the Python environment used by the project.

It does **not** perform analysis. Run it first when moving the repository to a new machine or cluster session.

The main workflow relies on:

- `geopandas` and `shapely` for vector spatial data;
- `osmnx` and `networkx` for the walking network;
- `rasterio` and `xarray` for raster/gridded data;
- `voxcity` for urban-form voxel preparation;
- `solweig` for thermal simulation.

```python
from pathlib import Path

# The notebooks_v02 folder is expected to sit directly inside the repository.
# If you open Jupyter from another working directory, set PROJECT_ROOT manually.
HERE = Path.cwd().resolve()
PROJECT_ROOT = HERE.parent if HERE.name == "notebooks_v02" else HERE

DATA_RAW = PROJECT_ROOT / "data" / "raw"
DATA_INTERIM = PROJECT_ROOT / "data" / "interim"
OUTPUTS = PROJECT_ROOT / "outputs"

for folder in [DATA_RAW, DATA_INTERIM, OUTPUTS]:
    folder.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
```

```python
import importlib

packages = [
    "geopandas",
    "matplotlib",
    "networkx",
    "numpy",
    "osmnx",
    "pandas",
    "rasterio",
    "xarray",
    "voxcity",
    "solweig",
]

for package in packages:
    try:
        module = importlib.import_module(package)
        version = getattr(module, "__version__", "version not exposed")
        print(f"{package:12s} OK   {version}")
    except Exception as exc:
        print(f"{package:12s} FAIL {exc}")
```

## Expected outcome

Every core package should import successfully. Package installation is intentionally kept outside the analytical notebooks so that the workflow remains reproducible and uncluttered.
